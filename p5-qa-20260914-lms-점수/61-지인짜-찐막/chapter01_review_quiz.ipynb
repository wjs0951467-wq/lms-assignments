튜터님 답:
1. 회원들의 이름과 나이 목록을(list) 준비한다.

3. 회원  한 명씩 순회하며 아래 내용을 진행한다.
	3.1현재 회원의 나이를 최대값 종이의 값과 비교
		3.2 만약 최대값보다 크면 최대값 종이의 값을 바꿈.

```python
#records
# 2. 종이 2장을 준비한다.(종이에 내용은 없다. 최대 최소 나이 적어야 함.) ==>0
max_age = 0
min_age = 1000
# 3. 회원  한 명씩 순회하며 아래 내용을 진행한다.
for rec in records:
	age = rec["Age"]
	# 3.1현재 회원의 나이를 최대값 종이의 값과 비교
	if age > max_age:
		# 3.2 만약 최대값보다 크면 최대값 종이의 값을 바꿈.
		max_age = age
		# 3.3 만약 최소값보다 작으면 최소값 종이의 값을 바꿈.
	if age < min_age:
		min_age = age
#4. 종이 내용을 보고
print("max_age : ", max_age)
print("min_age : ", min_age)
```

```python
old_date = records[0]["Signup_Date"]
recent_date = records[0]["Signup_Date"]

for rec in records:
	Signup_Date = rec["Signup_Date"]
	
	if Signup_Date < old_date:
		old_date = Signup_Date
	if Signup_Date > recent_date:
		recent_date = Signup_Date
print("old_date : ", old_date)
print("recent_date : ", recent_date)
```

```python
total_age = 0

for rec in records:
    age = rec["Age"]
    total_age += age

print("total_age : ", total_age)

customer_count = 0

for rec in records:
    customer_count = customer_count +1

print("customer_count :", customer_count)

print("average :", total_age/customer_count)
```
